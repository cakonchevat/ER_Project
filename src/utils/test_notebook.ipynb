```python
import numpy as np
print("Fraction predicted matches:", np.mean(df.pred_match))
print("Mean prob for predicted matches:", df.loc[df.pred_match==1, "prob_match"].mean())
print("Mean prob for predicted non-matches:", df.loc[df.pred_match==0, "prob_match"].mean())
```

```python
import matplotlib.pyplot as plt 
df.pred_match.hist()
plt.show()
```

```python
y_prob = df["prob_match"].astype(float)

display(df[["prob_match"]].describe(percentiles=[.5, .9, .95, .99]))
y_prob.hist(bins=50); plt.xlabel("prob_match"); plt.ylabel("count"); plt.title("Probability histogram"); plt.show()

mid = y_prob[(y_prob>=0.2)&(y_prob<=0.8)]
mid.hist(bins=40); plt.xlabel("prob_match (0.2–0.8)"); plt.ylabel("count"); plt.title("Mid-range zoom");
plt.show()
```

```python
# From 0.8 upwards we see very confident matches
# Whereas from 0.4 to 0.7 is an ambiguous area
# I will use 0.45 as a threshold and prioritise recall when making a decision and later cut (or connect)
# components (entities) through transitivity (clustering)
```

```python
gold_pairs = pd.read_csv("../../data/original/affiliationstrings_mapping.csv", header=None, names=["a", "b"])
```

```python
cluster_of = pred.set_index("node_id")["cluster_id"].to_dict()
```

```python
# Comparing how often gold-linked pairs fall in a same cluster
total_pairs = len(gold_pairs)
same_cluster = 0
missing_nodes = 0

for _, row in gold_pairs.iterrows():
    a, b = int(row["a"]), int(row["b"])
    if a not in cluster_of or b not in cluster_of:
        missing_nodes += 1
        continue
    if cluster_of[a] == cluster_of[b]:
        same_cluster += 1

precision_like = same_cluster / (total_pairs - missing_nodes)
coverage = (total_pairs - missing_nodes) / total_pairs

print(f"Gold pairs total: {total_pairs}")
print(f"Pairs found in predicted data: {total_pairs - missing_nodes} ({coverage:.1%})")
print(f"Pairs correctly grouped in same cluster: {same_cluster}")
print(f"→ Match rate (pairwise recall-like): {precision_like:.3f}")
```
